## 1. Project Overview

Customer churn occurs when customers stop purchasing products or discontinue their relationship with an e-commerce business. Understanding customer churn is important for identifying customers who may leave and developing suitable customer retention strategies.

This project focuses on analyzing e-commerce customer churn patterns, identifying customer characteristics associated with churn, and exploring factors related to customer retention. Python and its data analysis libraries will be used to inspect, clean, analyze, and visualize the dataset.


## 2. Business Problem

E-commerce businesses may experience customer loss because of poor customer satisfaction, reduced purchasing activity, service issues, or other factors. Losing customers can affect sales, revenue, and long-term business growth.

Therefore, analyzing customer churn data can help identify patterns in customer behavior and highlight factors associated with customer loss. The findings may support better-informed customer retention decisions.


## 3. Project Objectives

The main objectives of this project are:

1. To inspect and understand the e-commerce customer churn dataset.
2. To identify the overall customer churn pattern.
3. To examine customer characteristics associated with churn.

5. To explore the relationship between customer satisfaction and churn.

8. To provide data-driven insights that may support customer retention strategies.


## 4. Dataset Information

**Dataset Name:** E-Commerce Customer Churn Dataset

**Source:** Kaggle

**Dataset Link:** https://www.kaggle.com/datasets/datascikhan/e-commerce-customer-churn-2026

**Description:**

The dataset contains 48 columns related to customer information, purchasing behavior, satisfaction, churn status, marketing engagement, customer value, and retention activities. These variables will be used to explore customer churn patterns and identify characteristics associated with customer retention.

### Dataset Variables

The dataset includes the following categories of variables:

1. **Customer Information:** `churn_id`, `customer_id`, `customer_segment`, `customer_country`, `customer_city`
2. **Churn Information:** `is_churned`, `churn_flag`, `churn_date`, `days_since_churn`, `churn_reason`, `churn_category`
3. **Churn Risk Indicators:** `churn_risk_score`, `churn_probability`, `is_high_risk_customer`, `predicted_churn`, `churn_risk_category`, `churn_rate_indicator`
4. **Purchasing and Subscription Information:** `subscription_type`, `customer_age_months`, `total_spend_usd`, `avg_order_value_usd`, `num_purchases`, `days_since_last_purchase`
5. **Customer Satisfaction and Support:** `satisfaction_score`, `satisfaction_level`, `support_tickets`, `returns_count`, `support_interactions`, `last_support_channel`, `feedback_score`
6. **Customer Value and Engagement:** `customer_lifetime_value_usd`, `engagement_level`, `loyalty_score`, `customer_value_category`, `customer_health_score`, `customer_health_category`
7. **Marketing and Referrals:** `marketing_emails_sent`, `marketing_emails_opened`, `marketing_emails_clicked`, `marketing_conversions`, `social_media_engagement`, `referral_count`
8. **Retention and Revenue:** `prevention_method`, `customer_retention_cost_usd`, `recovery_revenue_usd`, `churn_revenue_impact_usd`, `retention_probability`
9. **Temporal Information:** `dataset_year`

The actual number of rows and the data types of the variables will be verified during the initial dataset inspection.
