```python
# Identify potential outliers using the IQR method

numerical_cols = df_clean.select_dtypes(include='number').columns

outlier_results = []

for col in numerical_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ).sum()

    if outlier_count > 0:
        outlier_results.append({
            'Column': col,
            'Potential Outliers': outlier_count
        })

outlier_df = pd.DataFrame(outlier_results)

print(outlier_df.to_string(index=False))
```

Outlier Detection Using the IQR Method

The Interquartile Range (IQR) method was used to identify potential outliers in the numerical columns.

The IQR is calculated as the difference between the third quartile (Q3) and the first quartile (Q1).

Values below Q1 - 1.5 × IQR or above Q3 + 1.5 × IQR are identified as potential outliers.

These observations were flagged for further investigation and were not automatically considered errors.

```python
# Investigate potential outliers

outlier_cols = outlier_df['Column'].tolist()

# Display descriptive statistics for all potential outlier columns
df_clean[outlier_cols].describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
).T
```

The potential outliers identified using the IQR method were further investigated by examining their descriptive statistics, value ranges, and relevant customer information.
