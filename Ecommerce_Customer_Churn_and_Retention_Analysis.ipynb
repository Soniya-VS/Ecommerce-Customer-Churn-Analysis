```python
# Identify potential outliers using the IQR method

numerical_cols = df_clean.select_dtypes(include='number').columns

outlier_results = []

for col in numerical_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ).sum()

    if outlier_count > 0:
        outlier_results.append({
            'Column': col,
            'Potential Outliers': outlier_count
        })

outlier_df = pd.DataFrame(outlier_results)

print(outlier_df.to_string(index=False))
```

Outlier Detection Using the IQR Method

The Interquartile Range (IQR) method was used to identify potential outliers in the numerical columns.

The IQR is calculated as the difference between the third quartile (Q3) and the first quartile (Q1).

Values below Q1 - 1.5 × IQR or above Q3 + 1.5 × IQR are identified as potential outliers.

These observations were flagged for further investigation and were not automatically considered errors.

```python
# Investigate potential outliers

outlier_cols = outlier_df['Column'].tolist()

# Display descriptive statistics for all potential outlier columns
df_clean[outlier_cols].describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
).T
```

The relationship between support interactions, purchase recency, and churn status was examined to identify whether customers with more support interactions and longer periods of purchase inactivity show higher churn rates.

```python
# Create a temporary copy for grouped analysis
df_multi = df_clean.copy()

# Categorize purchase recency
df_multi['recency_group'] = pd.cut(
    df_multi['days_since_last_purchase'],
    bins=[0, 30, 90, 180, 364],
    labels=['0-30 days', '31-90 days', '91-180 days', '181-364 days'],
    include_lowest=True
)

# Categorize support interactions
df_multi['support_group'] = pd.cut(
    df_multi['support_interactions'],
    bins=[-1, 5, 10, 15, float('inf')],
    labels=['0-5', '6-10', '11-15', 'Above 15']
)

# Calculate churn percentage
churn_heatmap = df_multi.pivot_table(
    index='support_group',
    columns='recency_group',
    values='is_churned',
    aggfunc=lambda x: (x == 'Yes').mean() * 100,
    observed=False
)

print("Churn Percentage by Support Interactions and Purchase Recency:")
print(churn_heatmap.round(2))

# Heatmap visualization
plt.figure(figsize=(9, 6))

sns.heatmap(
    churn_heatmap,
    annot=True,
    fmt='.2f',
    cmap='YlOrRd',
    linewidths=0.5,
    cbar_kws={'label': 'Churn Percentage (%)'}
)

plt.title('Churn Rate by Support Interactions and Purchase Recency')
plt.xlabel('Days Since Last Purchase')
plt.ylabel('Number of Support Interactions')
plt.tight_layout()
plt.show()
```

Interpretation
- Churn rates generally increase as the number of support interactions and days since the last purchase increase.
- Customers with 0–5 support interactions and 0–30 days since their last purchase have the lowest churn rate (33.11%).
- Customers with more than 15 support interactions and 181–364 days since their last purchase have the highest churn rate (57.47%).
- Customers with 11–15 support interactions show relatively high churn rates across all purchase recency categories.
- The results suggest that longer purchase inactivity combined with frequent support interactions is associated with higher customer churn.
- However, the pattern is not strictly increasing across every category.
